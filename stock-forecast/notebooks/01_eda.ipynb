## 2. Prices vs. returns

```python
wide_close = df.pivot(index="date", columns="ticker", values="close")
wide_ret = df.pivot(index="date", columns="ticker", values="log_ret")

fig, axes = plt.subplots(2, 1, figsize=(11, 7), sharex=True)
(wide_close / wide_close.iloc[0]).plot(ax=axes[0], logy=True, title="Price, rebased to 1 (log scale)")
wide_ret[config.TICKERS[0] if config.TICKERS[0] in wide_ret else wide_ret.columns[0]].plot(
    ax=axes[1], lw=0.6, title="Daily log return")
plt.tight_layout()
```
